```python
# Choose the relevant colum and Fill in the missing code

type_counts = df[''].value_counts()

# Create the bar chart
plt.figure(figsize=(10, 6))
type_counts.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title('')
plt.xlabel('')
plt.ylabel('')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

# **Exercise 3. Comparing Data for relationships, correlations and trends**

**Draw a scatter plot to check the correlation between 'Age' and the 'Fare', Write your comments here:**

```python
# Choose the relevant columns and fill in the missing code


# Create the scatter plot with regression line
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='', y='', color='teal', edgecolor='black', alpha=0.7)
sns.regplot(data=df, x='', y='', scatter=False, color='darkred', line_kws={'label':"Linear Fit"})

plt.title('')
plt.xlabel('')
plt.ylabel('')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
