## **Bar charts**

**Draw a Bar Chart to the number of passengers who 'Survived'. Remember Survived with value 0 means did not survive, with value 1 means survived. Give you comments on the chart here:**

The survival bar chart shows the tragic outcome of the Titanic disaster:

1. **More deaths than survivors**: Significantly more passengers (value 0) did not survive compared to those who survived (value 1)
2. **Survival rate**: Approximately 38% of passengers survived, while 62% perished in the disaster

The chart clearly illustrates the devastating impact of the Titanic sinking, with the majority of passengers losing their lives. This visualization helps quantify the human cost of the maritime disaster and shows the stark reality of survival rates during the tragedy.



```python
# Choose the relevant colum and Fill in the missing code

type_counts = df['Survived'].value_counts()

# Create the bar chart
plt.figure(figsize=(10, 6))
type_counts.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title('Number of Passengers by Survival Status')
plt.xlabel('Survival Status (0 = Did not survive, 1 = Survived)')
plt.ylabel('Number of Passengers')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

# **Exercise 3. Comparing Data for relationships, correlations and trends**

**Draw a scatter plot to check the correlation between 'Age' and the 'Fare', Write your comments here:**

The scatter plot between Age and Fare reveals:

- **Weak positive correlation**: The regression line shows a slight upward trend, indicating a weak positive relationship between age and fare
- **High variability**: Data points are widely scattered, suggesting fare is influenced by many factors beyond just age
- **Outliers present**: Several passengers paid very high fares regardless of age (upper outliers)
- **Age range**: Passengers span from infants to elderly, but fare distribution varies significantly within each age group
- **Economic factors**: Fare appears more related to passenger class and cabin location than age alone

**My Observation**

While there's a slight tendency for older passengers to pay higher fares, the relationship is weak, indicating that fare pricing was primarily based on other factors like passenger class, cabin location, and family size rather than age.

```python
# Choose the relevant columns and fill in the missing code


# Create the scatter plot with regression line
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Age', y='Fare', color='teal', edgecolor='black', alpha=0.7)
sns.regplot(data=df, x='Age', y='Fare', scatter=False, color='darkred', line_kws={'label':"Linear Fit"})

plt.title('Correlation between Age and Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
