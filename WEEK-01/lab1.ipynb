# EXERCISE - 05 ( IRIS DATASET )

```python
# LOADING THE IRIS DATASET

df = pd.read_csv('/content/iris.csv')
```

```python
# THIS CODE WILL RENAMES EACH OF THE COLUMNS WITH A MORE DESCRIPTIVE NAME FOR THE DATA THEY CONTAIN

df.rename(columns={
    "Type": "Plant Class",
    "PW": "Petal Width",
    "PL": "Petal Length",
    "SW": "Sepal Width",
    "SL": "Sepal Length"
})
```

```python
# PROVIDES THE FIRST 6 VALUES AND INFORMATION OF THE DATASET

(df.head(6))
```

```python
# PROVIDES THE LAST 3 VALUES AND THEIR INFORMATION OF THE DATASET

(df.tail(3))
```

```python
# PROVIDES THE STATESTICAL INFROMATION OF THE DATASET

df.describe()
```

```python
# PROVIDES ALL THE ROWS IN THE IRIS DATASET THAT HAVE A PW GREATER THAN 10 AND A PL GREATER THAN 50
df[(df['PW'] > 10) & (df['PL'] > 50)]
```

```python
# CREATES THE COPY OF THE 'TYPES' COLUMN AND NAME THE NEW COLUMN IN THE DATA FRAME 'SPECIES'

df['Species'] = df['Type'].copy()
```

```python
# CREATES A RECENTLY UPDATED DATA FRAME TO ANEW FILE IN OUT DATASETS FOLDER CALLED 'new_iris.csv'

df.to_csv("/content/new_iris.csv", index=False)
```
