**Give your interpretation of the results**

1.   List item

---------------------> Solution-01 <--------------------
1. **Skewness Interpretation**: If the skewness value is > 1 or < -1, the data is highly skewed. If skewness ≈ 0, the data is roughly normally distributed.





2.   List item

---------------------> Solution-02 <--------------------

2. **Filling Strategy**:
   - **Highly skewed data** (|skewness| > 1): Use **median** for filling missing values because median is robust to outliers and better represents the central tendency
   - **Normal data** (skewness ≈ 0): Use **mean** for filling missing values as it provides the best central measure
   - **Rule of thumb**: When mean and median are very different, the data is likely skewed and median should be preferred





```python
# You can now fill the Total_Units with your chosen method.
# Based on skewness analysis, if highly skewed use median, if normal use mean

# Check skewness first to decide the method
skewness = df['Total_Units'].skew()
print(f"Skewness: {skewness}")

if abs(skewness) > 1:
    # Highly skewed - use median
    fill_value = df['Total_Units'].median()
    print(f"Using median: {fill_value}")
else:
    # Roughly normal - use mean
    fill_value = df['Total_Units'].mean()
    print(f"Using mean: {fill_value}")

# Fill missing values
df['Total_Units'] = df['Total_Units'].fillna(fill_value)
```

```python
# Now check if there are missing values for Total_Units
print("Missing values in Total_Units after filling:", df['Total_Units'].isnull().sum())
print("Verification: All missing values filled successfully!")
```
